```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/netflix_cleaned.csv')
df['date_added'] = pd.to_datetime(df['date_added'])

# new column 1: the year each title was added (your Task 2 move!)
df['year_added'] = df['date_added'].dt.year

# new column 2: movie length as a pure number (only movies have it)
movies = df[df['type'] == 'Movie'].copy()
movies['minutes'] = movies['duration'].str.replace(' min', '').astype(int)
df = df.merge(movies[['show_id', 'minutes']], on='show_id', how='left')

df.to_csv('/content/drive/MyDrive/netflix_dashboard.csv', index=False)
print("Saved! Rows:", df.shape[0])
```
